# create new column - age_group

```python
labels = ["Young Adult", "Adult", "Middle Aged", "Senior"]
def get_age_group(age):
    if age <= 25:
        return "Young Adult"
    elif age < 35:
        return "Adult"
    elif age < 55:
        return "Middle Aged"
    return "Senior"
df['age_group'] = df['age'].apply(get_age_group)
```

```python
df[['age', 'age_group']]
```

# create new column - purchase_frequency_days

```python
df["frequency_of_purchases"].unique()
```

```python
mapping = {
    "Fortnightly" : 14,
    "Weekly" : 7,
    "Annually" : 365,
    "Quarterly" : 90,
    "Bi-Weekly" : 14,
    "Monthly":30,
    "Every 3 Months":90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(mapping)
```

# store to postgres 

```python
import pandas as pd
from sqlalchemy.engine import URL
from sqlalchemy import create_engine

# PostgreSQL connection details
username = "postgres"
password = "your_password"
host = "localhost"
port = "5432"
database = "customer_behaviour"

# Create connection
connection_url = URL.create(
    "postgresql+psycopg2",
    username=username,
    password=password,
    host=host,
    port=port,
    database=database
)

engine = create_engine(connection_url)

# Store DataFrame in PostgreSQL
df.to_sql(
    "customer",
    engine,
    if_exists="replace",   # replace / append / fail
    index=False
)

print("DataFrame successfully stored in PostgreSQL!")
```
